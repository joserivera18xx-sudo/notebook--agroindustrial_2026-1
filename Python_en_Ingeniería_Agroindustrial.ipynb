### ¿Qué es Python?

En términos sencillos, **Python** es un lenguaje de programación de alto nivel, lo que significa que su código es muy parecido al lenguaje humano (específicamente al inglés). Es famoso por ser **legible, versátil y gratuito**.

### ¿Por qué es importante en la Ingeniería Agroindustrial?

La agroindustria hoy en día no es solo tractores y silos; es **datos y precisión**. Python es clave por tres razones principales:

1. **Agricultura de Precisión:** Permite procesar datos de sensores de humedad, temperatura y nutrientes en tiempo real para tomar decisiones sobre el riego o la fertilización.
2. **Control de Calidad:** Se usa en visión artificial para clasificar frutas o granos por tamaño, color o defectos de forma automática.
3. **Optimización de Procesos:** Ayuda a calcular las rutas de transporte más cortas o a predecir cuánto tiempo durará un producto en anaquel basándose en variables ambientales.

### Tres Ejemplos de Código (Sencillos y Útiles)

Aquí tienes tres scripts básicos que podrías empezar a usar hoy mismo:

#### 1. Cálculo del Índice de Madurez (Relación Sólidos/Acidez)

Este código ayuda a determinar si una fruta está lista para la cosecha basándose en los Grados Brix y el porcentaje de acidez.

```
brix = 12.5
acidez = 0.8

indice_madurez = brix / acidez

print(f"El índice de madurez es: {indice_madurez}")

if indice_madurez > 15:
    print("Estado: Óptimo para cosecha/consumo.")
else:
    print("Estado: Aún le falta madurar.")
```

#### 2. Estimación de Evapotranspiración (Riego)

Un ejemplo muy simple para saber cuánta agua está perdiendo un cultivo de forma teórica simplificada.

```
# Datos diarios en mm
evaporacion_tanque = 6.2
coeficiente_cultivo = 0.85 # Depende de la etapa de la planta (Kc)

necesidad_riego = evaporacion_tanque * coeficiente_cultivo

print(f"La planta necesita {necesidad_riego} mm de agua hoy.")
```

#### 3. Control de Inventario de Materia Prima

Ideal para saber rápidamente si necesitas hacer un pedido de insumos.

```
insumos = {
    "Fertilizante NPK": 50, # kg
    "Semillas Maíz": 12,    # sacos
    "Plaguicida": 5         # litros
}

for nombre, cantidad in insumos.items():
    if cantidad < 10:
        print(f"¡ALERTA! El insumo {nombre} está bajo ({cantidad}). Reabastecer.")
    else:
        print(f"{nombre}: Stock suficiente.")
```
