```python
import math

# IDENTIFICACIÓN: [VICTOR JOSE TEHERAN RIVERA ]
# 1. ENTRADAS
radio = float(input("Ingrese el radio del silo (m): "))
altura = float(input("Ingrese la altura del silo (m): "))
densidad_maiz = 0.75

# 2. PROCESO
volumen = math.pi * (radio**2) * altura
toneladas = volumen * densidad_maiz

# 3. SALIDA
print(f"Capacidad total: {toneladas:.2f} toneladas de maíz")

# PRUEBA DE ESCRITORIO: Radio=3, Altura=10 -> Salida: ~212.06
```

```python
# IDENTIFICACIÓN: [VICTOR JOSE TEHERAN RIVERA ]
# 1. ENTRADAS
hectareas = float(input("Ingrese las hectáreas totales: "))
dosis_ha = float(input("Ingrese la dosis (L/ha): "))

# 2. PROCESO
litros_producto = hectareas * dosis_ha
litros_agua = litros_producto * 500

# 3. SALIDA
print(f"Producto necesario: {litros_producto} L")
print(f"Agua total necesaria: {litros_agua} L")

# PRUEBA DE ESCRITORIO: 5 ha, 2 L/ha -> Salida: 10L producto, 5000L agua
```

```python
# IDENTIFICACIÓN: [VICTOR TEHERAN RIBERA ]
# 1. ENTRADAS
peso_total = float(input("Ingrese el peso total de fruta (kg): "))
residuos = float(input("Ingrese el peso de los residuos (kg): "))

# 2. PROCESO
pulpa = peso_total - residuos
rendimiento = (pulpa / peso_total) * 100

# 3. SALIDA
print(f"Resultado: {pulpa} kg de pulpa ({rendimiento:.1f}% de rendimiento)")

# PRUEBA DE ESCRITORIO: 500kg total, 150kg residuos -> Salida: 350kg (70%)
```
